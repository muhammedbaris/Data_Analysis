```python
# 📊 Streaming KPI Data Reconstruction (2018–2024)
# ------------------------------------------------------------
# This notebook simulates revenue, subscriber, and churn data
# for five major streaming platforms using public source estimates.
#
# Platforms analyzed: Netflix (NFLX), Disney+ (DIS), Amazon Prime Video (AMZN),
# Warner Bros Discovery (WBD), and Apple TV+ (AAPL).
#
# Source assumptions:
# - Revenue: Macrotrends, Business of Apps, Statista (streaming segments)
# - Subscribers: Wikipedia, DemandSage, Business of Apps
# - Churn: Antenna, Churnkey (industry average estimates)
```

```python
# ----------------------------------------
# 1️⃣ Import libraries and set base structure
# ----------------------------------------
import pandas as pd

# Define the streaming companies by abbreviation
companies = ['NFLX', 'DIS', 'AMZN', 'WBD', 'AAPL']

# Define the analysis period (2018 to 2024 inclusive)
years = list(range(2018, 2025))

# Create a multi-index of all company-year combinations
index = pd.MultiIndex.from_product([companies, years], names=["Company", "Year"])
```

```python
# Convert to flat DataFrame for merging data later
df_base = pd.DataFrame(index=index).reset_index()

# -------------------------------------------------
# 2️⃣ Add Revenue (in Billions USD, streaming only)
# -------------------------------------------------
# Revenue values are manually entered based on public reverse-engineered sources.
revenue_data = {
    'NFLX': [15.79, 20.16, 24.99, 29.70, 31.62, 33.72, 39.00],  # Macrotrends + Netflix IR
    'DIS':  [None, None, 2.80, 5.20, 7.40, 8.40, 10.40],          # Disney+ launch in late 2019
    'AMZN': [3.57, 4.10, 4.63, 5.16, 14.60, 14.60, 31.00],        # Prime Video estimates
    'WBD':  [None, None, 6.09, 7.70, 8.00, 8.00, 9.00],           # HBO Max estimates (Business of Apps)
    'AAPL': [None, None, 0.50, 0.70, 1.00, 1.00, 1.20]            # Apple TV+ niche service
}

# Transform revenue dictionary into long format rows
revenue_rows = []
for company, values in revenue_data.items():
    for year, value in zip(years, values):
        revenue_rows.append({'Company': company, 'Year': year, 'Revenue_B_USD': value})

        # Create a revenue DataFrame
df_revenue = pd.DataFrame(revenue_rows)

# Merge revenue into base dataset
df_combined = pd.merge(df_base, df_revenue, on=['Company', 'Year'], how='left')
```

```python
# -------------------------------------------------
# 3️⃣ Add Subscribers (in Millions)
# -------------------------------------------------
# Subscriber data sourced from DemandSage, Wikipedia, Backlinko, etc.
subscribers_data = {
    'NFLX': [139.3, 167.1, 203.7, 221.8, 230.7, 260.3, 301.6],
    'DIS':  [None, 10.0, 73.7, 118.1, 152.1, 146.1, 124.6],
    'AMZN': [125.0, 150.0, 200.0, 161.7, 168.3, 174.9, 180.1],
    'WBD':  [None, None, 55.6, 67.0, 81.2, 95.8, 117.0],
    'AAPL': [None, 5.7, 33.6, 48.0, 55.0, 40.0, 45.0]
}

# Convert subscriber data into long format
subs_rows = []
for company, subs in subscribers_data.items():
    for year, value in zip(years, subs):
        subs_rows.append({'Company': company, 'Year': year, 'Subscribers_M': value})

        # Create DataFrame and merge into main structure
df_subs = pd.DataFrame(subs_rows)
df_combined = pd.merge(df_combined, df_subs, on=['Company', 'Year'], how='left')
```

```python
# -------------------------------------------------
# 4️⃣ Add Annual Churn Rates (% of users who leave)
# -------------------------------------------------
# Churn estimates are approximated by platform maturity and business model
churn_data = {
    'NFLX': [7.5, 7.3, 7.1, 6.9, 6.5, 6.3, 6.1],               # Netflix: low churn, mature platform
    'DIS':  [None, 45.0, 43.5, 42.1, 43.3, 44.0, 42.7],         # Disney+: higher churn, family-centric
    'AMZN': [8.0, 8.0, 7.5, 7.0, 6.7, 6.5, 6.3],               # Amazon Prime: very low churn (bundled)
    'WBD':  [None, None, 55.0, 53.0, 52.0, 50.5, 49.0],         # HBO Max: high churn but declining
    'AAPL': [None, 57.0, 54.0, 51.5, 49.3, 48.2, 45.9]          # Apple TV+: highest churn (free trial heavy)
}

churn_rows = []
for company, churns in churn_data.items():
    for year, value in zip(years, churns):
        churn_rows.append({'Company': company, 'Year': year, 'Annual_Churn': value})

        # Create churn DataFrame and merge it in
df_churn = pd.DataFrame(churn_rows)
df_combined = pd.merge(df_combined, df_churn, on=['Company', 'Year'], how='left')
```
